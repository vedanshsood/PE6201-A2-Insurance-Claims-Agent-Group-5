```python
# ============================================================
# SAFE ASSIGNMENT REPLACEMENT
#
# This replaces the COMPLETE existing EXTRA_* assignment.
# Therefore Vedansh's existing records are NOT duplicated.
# ============================================================

def replace_assignment(source, variable_name, new_assignment):
    tree = ast.parse(source)

    target = None

    for node in tree.body:
        if isinstance(node, ast.Assign):
            for target_node in node.targets:
                if isinstance(target_node, ast.Name) and target_node.id == variable_name:
                    target = node
                    break

        if target is not None:
            break

    if target is None:
        raise ValueError(f"Could not find assignment for {variable_name}")

    lines = source.splitlines(keepends=True)

    start = target.lineno - 1
    end = target.end_lineno

    replacement_lines = [new_assignment + "\n"]

    updated = (
        lines[:start]
        + replacement_lines
        + lines[end:]
    )

    return "".join(updated)


# Replace one section at a time.
# Re-parse after each replacement so line numbers remain correct.

replacements = {
    "EXTRA_PROCEDURES": extra_procedures,
    "EXTRA_POLICIES": extra_policies,
    "EXTRA_MEMBERS": extra_members,
    "EXTRA_PREAUTHORISATIONS": extra_preauth,
    "EXTRA_CLAIMS": extra_claims,
    "EXTRA_DECIDED": extra_decided,
}

for variable_name, replacement in replacements.items():
    text = replace_assignment(
        text,
        variable_name,
        replacement
    )
    print("✓ Replaced:", variable_name)


# We don't currently need extra hospitals or extra document rules.
# Existing values remain unchanged.

fixture_path.write_text(text)

print("\n==============================================")
print("✓ TEAM FIXTURE SETUP WRITTEN SUCCESSFULLY")
print("==============================================")
print("Vedansh cases retained: 5")
print("New teammate cases added: 22")
print("Total team-created cases: 27")
```

```python
# ============================================================
# BUILD OUTCOME LOOKUP FOR ALL 27 TEAM CASES
# ============================================================

our_outcomes = {
    o["case_id"]: o
    for o in outcomes
    if o["case_id"] in our_ids
}


# ============================================================
# BUILD SUMMARY TABLE
# ============================================================

rows = []

for claim in sorted(our_claims, key=lambda x: x["claim_id"]):

    outcome = our_outcomes[claim["claim_id"]]

    cid = claim["claim_id"]

    if cid.startswith("CLM-V"):
        owner = "Vedansh"
    elif cid.startswith("CLM-Q"):
        owner = "Qian Ruoqi"
    elif cid.startswith("CLM-C"):
        owner = "Chen Baiyi"
    elif cid.startswith("CLM-Z"):
        owner = "Zhang Qizhi"
    elif cid.startswith("CLM-F"):
        owner = "Fang Hioieng"
    elif cid.startswith("CLM-Y"):
        owner = "Yang Yichen"
    else:
        owner = "Unknown"

    rows.append({
        "Owner": owner,
        "Case ID": claim["claim_id"],
        "Member": claim["member_id"],
        "Date": claim["date_of_service"],
        "Claim Total": sum(
            x["amount"]
            for x in claim["lines"]
        ),
        "Expected Outcome": outcome["expected_decision"],
        "Test Family": outcome["family"]
    })


eval_df = pd.DataFrame(rows)


# ============================================================
# SORT IN TEAM ORDER
# ============================================================

owner_order = [
    "Vedansh",
    "Qian Ruoqi",
    "Chen Baiyi",
    "Zhang Qizhi",
    "Fang Hioieng",
    "Yang Yichen"
]

eval_df["Owner"] = pd.Categorical(
    eval_df["Owner"],
    categories=owner_order,
    ordered=True
)

eval_df = eval_df.sort_values(
    ["Owner", "Case ID"]
).reset_index(drop=True)


# ============================================================
# DISPLAY
# ============================================================

display(eval_df)
```
